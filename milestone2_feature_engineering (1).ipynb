```python
import pandas as pd
import numpy as np
df = pd.read_excel("economic_data_cleaned.xlsx")
df.columns = df.columns.str.strip()
df['Date'] = pd.date_range(start='2019-01-01', periods=len(df), freq='M')
df = df.sort_values('Date').reset_index(drop=True)
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Quarter'] = df['Date'].dt.quarter
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
exclude_cols = ['Year', 'Month', 'Quarter']
value_cols = [col for col in numeric_cols if col not in exclude_cols]
for col in value_cols:
    df[f'{col}_lag1'] = df[col].shift(1)
for col in value_cols:
    df[f'{col}_rolling3'] = df[col].rolling(3).mean()
df = df.dropna().reset_index(drop=True)
df.to_csv("economic_data_feature_engineered.csv", index=False)
print("✅ Feature Engineering Completed Successfully!")
```
